```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("../data/raw/dataset.csv")
df.head()
```

```python
btc = df[df["crypto_name"]=="Bitcoin"]
plt.plot(pd.to_datetime(btc["date"]), btc["close"])
plt.title("Bitcoin Price Trend")
plt.show()
```

```python
sns.heatmap(df[["open","high","low","close","volume","marketCap"]].corr(), annot=True)
plt.show()
```

```python
returns = btc["close"].pct_change()
returns.rolling(7).std().hist(bins=50)
plt.title("Volatility Distribution")
plt.show()
```

```python
vol = returns.rolling(7).std()
plt.scatter(btc["volume"], vol, alpha=0.3)
plt.xlabel("Volume")
plt.ylabel("Volatility")
plt.show()
```
